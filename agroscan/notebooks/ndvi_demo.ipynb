```python
import numpy as np
import matplotlib.pyplot as plt

# Synthetic red/NIR bands (replace with real data)
red = np.random.uniform(0.1, 0.6, (64, 64))
nir = np.random.uniform(0.2, 0.9, (64, 64))

ndvi = (nir - red) / (nir + red + 1e-6)
stress = np.clip(1 - ndvi, 0, 1)

fig, axs = plt.subplots(1, 2, figsize=(8, 4))
im0 = axs[0].imshow(ndvi, cmap='Greens', vmin=0, vmax=1); axs[0].set_title('NDVI')
im1 = axs[1].imshow(stress, cmap='OrRd', vmin=0, vmax=1); axs[1].set_title('Stress score')
for ax in axs: ax.axis('off')
fig.colorbar(im0, ax=axs[0], fraction=0.046, pad=0.04)
fig.colorbar(im1, ax=axs[1], fraction=0.046, pad=0.04)
plt.show()
```
